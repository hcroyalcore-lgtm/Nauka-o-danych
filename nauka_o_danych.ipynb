```python
import pandas as pd

dane = {
    "Klient": ["Anna", "Bartek", "Celina", "Dawid", "Ewa"],
    "Miasto": ["Warszawa", "Kraków", "Warszawa", "Gdańsk", "Kraków"],
    "Wiek": [28, 34, 22, 45, 29],
    "Kwota_zakupu": [120.50, 450.00, 89.90, 310.20, 215.00],
    "Status_VIP": [False, True, False, True, False],
}

df = pd.DataFrame(dane)
print(df)
```

```python
df.groupby("Status_VIP").agg({"Wiek": "mean", "Kwota_zakupu": "sum"})
```

```python
srednia_na_miasto = df.groupby("Miasto")["Kwota_zakupu"].mean()
display(srednia_na_miasto)
```

```python
df.groupby("Status_VIP").agg({"Wiek": "mean", "Kwota_zakupu": "sum"})
```

```python
df.groupby("Miasto")["Kwota_zakupu"].agg(["mean", "count"])
```

```python
df.groupby("Miasto")["Kwota_zakupu"].agg(["mean", "count", "max"])
```

```python
df.groupby("Miasto")["Kwota_zakupu"].agg([
    "mean",
    "count",
    "max",
    ("roznica_max_min", lambda x: x.max() - x.min())
])
```

```python
# 1. Procentowy udział klientów VIP w każdym mieście
udzial_vip = df.groupby("Miasto")["Status_VIP"].agg([
    ("procent_VIP", lambda x: (x.sum() / len(x)) * 100)
])
print("1. Procent VIP-ów w miastach:")
display(udzial_vip)

# 2. Specyficzne kwantyle (np. 90. percentyl kwoty zakupu)
percentyl_90 = df.groupby("Miasto")["Kwota_zakupu"].agg([
    ("90_percentyl", lambda x: x.quantile(0.9))
])
print("\n2. 90. percentyl kwoty zakupów:")
display(percentyl_90)

# 3. Zliczanie unikalnych wartości (odpowiednik SQL: COUNT DISTINCT)
unikalne_zakupy = df.groupby("Miasto")["Kwota_zakupu"].agg([
    ("liczba_unikalnych_kwot", lambda x: x.nunique())
])
print("\n3. Liczba unikalnych kwot zakupów:")
display(unikalne_zakupy)
```
